## Robustness Runs

The consolidated experiment tables distinguish fixed-latent forest repeats from full-pipeline retraining / split repeats. Global score rankings can remain highly correlated while the exact thresholded candidate set changes substantially.

For the canonical v3 fixed-latent forest repeats, Jaccard overlap with the primary flag set was 0.667 and 0.727. Full-pipeline retraining and source-split checks were less stable (primary-v3 Jaccard 0.150 and 0.294 respectively), despite high global rank correlations. The v7 seed-repeat comparison was also weaker than v3 (Jaccard 0.417).

**Interpretation:** Strong global ranking correlation does not imply stable final candidate membership. The exact flagged set is sensitive to representation and split changes, and this limitation is disclosed rather than hidden.
