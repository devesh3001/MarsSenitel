## Iteration Summary

| Version | Change | Symptom | Diagnosis | Fix | Outcome |
|---------|--------|---------|-----------|-----|---------|
| v1 | MSE baseline, 128d | Fine texture blurred | MSE favours smoothness | Establish baseline | Val MSE 0.002694, SSIM 0.661 |
| v2 | + SSIM loss | Structural smoothing persists | Need perceptual term | Add 0.1×(1−SSIM) | SSIM → 0.672, MSE +1.7% |
| v3 | 128d → 256d latent | Capacity ceiling | Underfitting | Expand latent dim | MSE 0.002599, SSIM 0.674 |
| Threshold | Boxplot rejected, GMM adopted | Fence exceeded all scores | Multimodal distribution | GMM BIC + bootstrap | Stable 95% CI threshold |
| v4 | Contrast normalization | Bright crops dominate | Illumination bias | CLAHE preprocessing | Borders became prominent; not promoted |
| v5 | Footprint masking | Border response persists | Near-black boundary artifacts | Mask periphery | 42 flags (too many); Jaccard collapsed |
| v6 | Fixed near-black mask | Strip artifacts from v5 | Fill method introduced stripes | Nearest-value fill | Defect response failed; negative result |
| v7 | + Gradient loss | Edge blurring unresolved | MSE+SSIM insufficient for ridges | Add 0.1×Gradient | MSE −1.68%; promoted to final |
| v8 | Flip/rotation augmentation | Orientation sensitivity | Test augmentation | 4× augmented training | Zero image-only flags; not promoted |
