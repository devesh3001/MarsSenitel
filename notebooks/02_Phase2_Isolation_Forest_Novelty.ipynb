## Phase 2.2 — Statistical Thresholding

**Method:** Gaussian Mixture Model with BIC model selection
- Fit GMM with 1–4 components to calibration split novelty scores
- Select minimum-BIC model
- **Threshold T = max_j(μⱼ + 3σⱼ)** — 3-sigma envelope beyond every fitted population
- 100 source-group bootstrap resamples → 95% CI

**Why not a fixed top-N cutoff?** Zero flags is an explicitly accepted outcome.
**Why not the adjusted boxplot?** It was tried first — rejected because the fence exceeded *every* score (multimodal distribution).

Final threshold: **T = 0.53917** (95% CI: 0.53261 – 0.55542)
