## Phase 1.1 — Architecture Design

5-stage Convolutional Autoencoder trained entirely from scratch.

| Layer | In Channels | Out Channels | Stride | Output Size |
|-------|-------------|--------------|--------|-------------|
| Enc-1 | 1 | 16 | 2 | 114×114 |
| Enc-2 | 16 | 32 | 2 | 57×57 |
| Enc-3 | 32 | 64 | 2 | 29×29 |
| Enc-4 | 64 | 96 | 2 | 15×15 |
| Enc-5 | 96 | 128 | 2 | 8×8 |
| Linear | 8192 | 256 | — | Latent z |

Each encoder block: `Conv2d(stride=2) → GroupNorm(8) → SiLU`
Decoder uses **resize-convolution** (bilinear upsample + Conv2d) — no transposed convolutions, preventing checkerboard artifacts.
